```python
# 1. Cost overrun percentage
df['Cost Overrun (%)'] = (
    (df['Revised Cost (in cr.)'] - df['Original Cost (in cr.)'])
    / df['Original Cost (in cr.)']
) * 100

# 2. Delay in days
df['Delay Days'] = (
    df['Revised Date of Commissioning']
    - df['Original Date of Commissioning']
).dt.days

# 3. Cost overrun flag
df['Cost Overrun Flag'] = (
    df['Cost Overrun (%)'] > 0
).astype(int)

# 4. Delay flag
df['Delay Flag'] = (
    df['Delay Days'] > 0
).astype(int)
```

```python
# Features for Cost Overrun prediction
features = [
    'Sector Name',
    'Line Ministry',
    'Implementing Agency',
    'Original Cost (in cr.)',
    'Expenditure (in cr.)',
    'Physical Progress (in %)',
    'Planned Duration (days)',
    'Expenditure Ratio (%)',
    'Expenditure-Progress Gap (%)',
    'Cost Base Anomaly'
]

X = df[features].copy()
y = df['Cost Overrun Flag'].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)
print("\nTarget distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)


# Separate categorical/numerical features
# -----------------------------

categorical_features = [
    'Sector Name',
    'Line Ministry',
    'Implementing Agency'
]

numerical_features = [
    'Original Cost (in cr.)',
    'Expenditure (in cr.)',
    'Physical Progress (in %)',
    'Planned Duration (days)',
    'Expenditure Ratio (%)',
    'Expenditure-Progress Gap (%)',
    'Cost Base Anomaly'
]

# -----------------------------
# Preprocessing
# -----------------------------

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median'))
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(
        handle_unknown='ignore',
        sparse_output=False
    ))
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numerical_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)

# -----------------------------
# Train/Test Split
# -----------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
# ==============================
# STEP 4: COST OVERRUN MODEL
# ==============================

# Create pipeline
cost_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        class_weight='balanced',
        max_depth=12,
        min_samples_split=5,
        n_jobs=-1
    ))
])

# Train model
cost_model.fit(X_train, y_train)

print("Cost Overrun Model trained successfull")
```

```python
# ==============================
# MODEL EVALUATION
# ==============================

y_pred = cost_model.predict(X_test)
y_prob = cost_model.predict_proba(X_test)[:, 1]

print("Accuracy:", accuracy_score(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nROC-AUC Score:")
print(roc_auc_score(y_test, y_prob))
```

```python
# ==========================================
# STEP 5: FEATURE IMPORTANCE
# ==========================================

# Get trained Random Forest
rf_model = cost_model.named_steps['classifier']

# Get transformed feature names
feature_names = cost_model.named_steps[
    'preprocessor'
].get_feature_names_out()

# Get importance
importances = rf_model.feature_importances_

# Create dataframe
importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importances
})

# Sort
importance_df = importance_df.sort_values(
    by='Importance',
    ascending=False
)

print(importance_df.head(20))
```

```python
# ==========================================
# STEP 6: PREPARE DELAY MODEL
# ==========================================

delay_features = [
    'Sector Name',
    'Line Ministry',
    'Implementing Agency',
    'Original Cost (in cr.)',
    'Expenditure (in cr.)',
    'Physical Progress (in %)',
    'Planned Duration (days)',
    'Expenditure Ratio (%)',
    'Expenditure-Progress Gap (%)',
    'Cost Base Anomaly'
]

# Keep only projects with known delay outcome
delay_df = df.dropna(subset=['Delay Flag']).copy()

X_delay = delay_df[delay_features].copy()
y_delay = delay_df['Delay Flag'].astype(int)

print("Delay dataset shape:", X_delay.shape)

print("\nDelay target distribution:")
print(y_delay.value_counts())

print("\nDelay target percentage:")
print(y_delay.value_counts(normalize=True) * 100)
```

```python
# ==========================================
# STEP 7: DELAY PREDICTION MODEL
# ==========================================

from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

# Split delay data
X_delay_train, X_delay_test, y_delay_train, y_delay_test = train_test_split(
    X_delay,
    y_delay,
    test_size=0.20,
    random_state=42,
    stratify=y_delay
)

print("Training data:", X_delay_train.shape)
print("Testing data:", X_delay_test.shape)


# Create Delay Model
delay_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        class_weight='balanced',
        max_depth=12,
        min_samples_split=5,
        n_jobs=-1
    ))
])


# Train
delay_model.fit(X_delay_train, y_delay_train)

print("\nDelay Prediction Model trained successfully!")
```

```python
# ==========================================
# STEP 8: DELAY MODEL EVALUATION
# ==========================================

y_delay_pred = delay_model.predict(X_delay_test)

y_delay_prob = delay_model.predict_proba(
    X_delay_test
)[:, 1]


print("Accuracy:",
      accuracy_score(y_delay_test, y_delay_pred))

print("\nClassification Report:")
print(
    classification_report(
        y_delay_test,
        y_delay_pred
    )
)

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_delay_test,
        y_delay_pred
    )
)

print("\nROC-AUC Score:")
print(
    roc_auc_score(
        y_delay_test,
        y_delay_prob
    )
)
```

```python
# ==========================================
# STEP 9: OVERALL PROJECT RISK SCORE
# ==========================================

def calculate_risk_score(cost_probability, delay_probability):

    # Convert probabilities to percentages
    cost_risk = cost_probability * 100
    delay_risk = delay_probability * 100

    # Equal weighting
    overall_score = (
        0.50 * cost_risk +
        0.50 * delay_risk
    )

    # Risk category
    if overall_score < 30:
        risk_level = "LOW"
    elif overall_score < 60:
        risk_level = "MEDIUM"
    else:
        risk_level = "HIGH"

    return round(overall_score, 2), risk_level
```

```python
# Example
cost_probability = 0.82
delay_probability = 0.74

score, level = calculate_risk_score(
    cost_probability,
    delay_probability
)

print("Cost Risk:", cost_probability * 100, "%")
print("Delay Risk:", delay_probability * 100, "%")
print("Overall Risk Score:", score)
print("Risk Level:", level)
```

```python
# ==========================================
# EARLY WARNING SYSTEM
# ==========================================

def generate_warning(cost_probability, delay_probability):

    score, level = calculate_risk_score(
        cost_probability,
        delay_probability
    )

    if level == "HIGH":
        warning = "IMMEDIATE ATTENTION REQUIRED"

    elif level == "MEDIUM":
        warning = "CLOSE MONITORING REQUIRED"

    else:
        warning = "PROJECT APPEARS STABLE"

    return score, level, warning
```

```python
# ==========================================
# STEP 11: COMPLETE PROJECT RISK PREDICTION
# ==========================================

import pandas as pd
import numpy as np
import joblib

# Load saved models
cost_model = joblib.load("cost_model.pkl")
delay_model = joblib.load("delay_model.pkl")


def predict_project_risk(project):

    # -------------------------------
    # 1. Create input dataframe
    # -------------------------------
    input_data = pd.DataFrame([project])

    # -------------------------------
    # 2. Cost Overrun Prediction
    # -------------------------------
    cost_probability = cost_model.predict_proba(
        input_data
    )[0][1]

    cost_prediction = cost_model.predict(
        input_data
    )[0]

    # -------------------------------
    # 3. Delay Prediction
    # -------------------------------
    delay_probability = delay_model.predict_proba(
        input_data
    )[0][1]

    delay_prediction = delay_model.predict(
        input_data
    )[0]

    # -------------------------------
    # 4. Convert to percentage
    # -------------------------------
    cost_risk = cost_probability * 100
    delay_risk = delay_probability * 100

    # -------------------------------
    # 5. Overall Risk Score
    # -------------------------------
    overall_score = (
        0.50 * cost_risk +
        0.50 * delay_risk
    )

    # -------------------------------
    # 6. Risk Level
    # -------------------------------
    if overall_score < 30:
        risk_level = "LOW"
        warning = "PROJECT APPEARS STABLE"

    elif overall_score < 60:
        risk_level = "MEDIUM"
        warning = "CLOSE MONITORING REQUIRED"

    else:
        risk_level = "HIGH"
        warning = "IMMEDIATE ATTENTION REQUIRED"

    # -------------------------------
    # 7. Risk Factors
    # -------------------------------
    risk_factors = []

    if project["Expenditure Ratio (%)"] > 100:
        risk_factors.append(
            "Expenditure has exceeded the project cost base"
        )

    if project["Expenditure-Progress Gap (%)"] > 20:
        risk_factors.append(
            "High expenditure-progress gap"
        )

    if project["Physical Progress (in %)"] < 50:
        risk_factors.append(
            "Low physical progress"
        )

    if project["Planned Duration (days)"] > 2000:
        risk_factors.append(
            "Long planned project duration"
        )

    if project["Revised Cost (in cr.)"] > project["Original Cost (in cr.)"]:
        risk_factors.append(
            "Revised cost is higher than original cost"
        )

    if not risk_factors:
        risk_factors.append(
            "No major threshold-based risk factor detected"
        )

    # -------------------------------
    # 8. Return results
    # -------------------------------
    return {
        "Cost Overrun Probability": round(cost_risk, 2),
        "Cost Overrun Prediction": (
            "YES" if cost_prediction == 1 else "NO"
        ),

        "Delay Probability": round(delay_risk, 2),
        "Delay Prediction": (
            "YES" if delay_prediction == 1 else "NO"
        ),

        "Overall Risk Score": round(overall_score, 2),
        "Risk Level": risk_level,
        "Early Warning": warning,
        "Risk Factors": risk_factors
    }

print("Prediction function created successfully!")
```

```python
# ==========================================
# FINAL PROJECT RISK PREDICTION FUNCTION
# ==========================================

def predict_project_risk(project):

    input_data = pd.DataFrame([project])

    # --------------------------------------
    # COST OVERRUN
    # --------------------------------------

    cost_probability = cost_model.predict_proba(
        input_data
    )[0][1]

    cost_prediction = cost_model.predict(
        input_data
    )[0]

    # --------------------------------------
    # DELAY
    # --------------------------------------

    delay_probability = delay_model.predict_proba(
        input_data
    )[0][1]

    delay_prediction = delay_model.predict(
        input_data
    )[0]

    # --------------------------------------
    # RISK %
    # --------------------------------------

    cost_risk = cost_probability * 100
    delay_risk = delay_probability * 100

    # --------------------------------------
    # OVERALL RISK
    # --------------------------------------

    overall_score = (
        0.50 * cost_risk +
        0.50 * delay_risk
    )

    # --------------------------------------
    # RISK LEVEL
    # --------------------------------------

    if overall_score < 30:

        risk_level = "LOW"
        warning = "PROJECT APPEARS STABLE"

    elif overall_score < 60:

        risk_level = "MEDIUM"
        warning = "CLOSE MONITORING REQUIRED"

    else:

        risk_level = "HIGH"
        warning = "IMMEDIATE ATTENTION REQUIRED"

    # --------------------------------------
    # RISK FACTORS
    # --------------------------------------

    risk_factors = []

    if project["Expenditure Ratio (%)"] > 100:
        risk_factors.append(
            "Expenditure has exceeded the cost base"
        )

    if project["Expenditure-Progress Gap (%)"] > 20:
        risk_factors.append(
            "High expenditure-progress gap"
        )

    if project["Physical Progress (in %)"] < 50:
        risk_factors.append(
            "Low physical progress"
        )

    if project["Planned Duration (days)"] > 2000:
        risk_factors.append(
            "Long planned project duration"
        )

    if project["Cost Base Anomaly"] == 1:
        risk_factors.append(
            "Abnormal cost base detected"
        )

    if not risk_factors:
        risk_factors.append(
            "No major threshold-based risk factor detected"
        )

    # --------------------------------------
    # RETURN
    # --------------------------------------

    return {

        "Cost Overrun Probability":
            round(cost_risk, 2),

        "Cost Overrun Prediction":
            "YES" if cost_prediction == 1 else "NO",

        "Delay Probability":
            round(delay_risk, 2),

        "Delay Prediction":
            "YES" if delay_prediction == 1 else "NO",

        "Overall Risk Score":
            round(overall_score, 2),

        "Risk Level":
            risk_level,

        "Early Warning":
            warning,

        "Risk Factors":
            risk_factors
    }

print("Final prediction function loaded successfully!")
```

```python
# ==========================================
# STEP 9: CLEAN SHAP FEATURE NAMES
# ==========================================

def clean_feature_name(feature):
    
    # Numerical features
    if feature.startswith("num__"):
        name = feature.replace("num__", "")
        
        replacements = {
            "Original Cost (in cr.)": "Original Cost",
            "Expenditure (in cr.)": "Expenditure",
            "Physical Progress (in %)": "Physical Progress",
            "Planned Duration (days)": "Planned Duration",
            "Expenditure Ratio (%)": "Expenditure Ratio",
            "Expenditure-Progress Gap (%)": "Expenditure-Progress Gap",
            "Cost Base Anomaly": "Cost Base Anomaly"
        }
        
        return replacements.get(name, name)
    
    # Categorical features
    if feature.startswith("cat__"):
        name = feature.replace("cat__", "")
        
        if "_" in name:
            column, value = name.split("_", 1)
            return f"{column}: {value}"
        
        return name
    
    return feature


# Apply cleaning
cost_shap_df["Clean Feature"] = cost_shap_df["Feature"].apply(
    clean_feature_name
)

delay_shap_df["Clean Feature"] = delay_shap_df["Feature"].apply(
    clean_feature_name
)

# Top 10
cost_top_clean = cost_shap_df.sort_values(
    "Importance", ascending=False
).head(10)

delay_top_clean = delay_shap_df.sort_values(
    "Importance", ascending=False
).head(10)


print("====================================")
print("     CLEAN COST SHAP FACTORS")
print("====================================")

print(
    cost_top_clean[
        ["Clean Feature", "SHAP Value"]
    ].to_string(index=False)
)


print("\n====================================")
print("     CLEAN DELAY SHAP FACTORS")
print("====================================")

print(
    delay_top_clean[
        ["Clean Feature", "SHAP Value"]
    ].to_string(index=False)
)
```

```python
# ==========================================
# STEP 10: SHAP RISK EXPLANATION FUNCTION
# ==========================================

def get_shap_explanation(shap_df, top_n=5):

    # Factors increasing risk
    risk_increasing = (
        shap_df[shap_df["SHAP Value"] > 0]
        .sort_values("SHAP Value", ascending=False)
        .head(top_n)
    )

    # Factors reducing risk
    risk_reducing = (
        shap_df[shap_df["SHAP Value"] < 0]
        .sort_values("SHAP Value", ascending=True)
        .head(top_n)
    )

    return risk_increasing, risk_reducing


# Cost explanation
cost_increasing, cost_reducing = get_shap_explanation(
    cost_shap_df
)

# Delay explanation
delay_increasing, delay_reducing = get_shap_explanation(
    delay_shap_df
)


print("====================================")
print("       COST RISK INCREASING")
print("====================================")

print(
    cost_increasing[
        ["Clean Feature", "SHAP Value"]
    ].to_string(index=False)
)


print("\n====================================")
print("       COST RISK REDUCING")
print("====================================")

print(
    cost_reducing[
        ["Clean Feature", "SHAP Value"]
    ].to_string(index=False)
)


print("\n====================================")
print("       DELAY RISK INCREASING")
print("====================================")

print(
    delay_increasing[
        ["Clean Feature", "SHAP Value"]
    ].to_string(index=False)
)


print("\n====================================")
print("       DELAY RISK REDUCING")
print("====================================")

print(
    delay_reducing[
        ["Clean Feature", "SHAP Value"]
    ].to_string(index=False)
)
```

```python
# ==========================================
# STEP 11: GENERAL SHAP EXPLANATION
# ==========================================

def explain_project(project):

    # Convert project dictionary to DataFrame
    input_data = pd.DataFrame([project])

    # ---------- COST ----------
    cost_transformed = cost_model.named_steps[
        'preprocessor'
    ].transform(input_data)

    if hasattr(cost_transformed, "toarray"):
        cost_transformed = cost_transformed.toarray()

    cost_transformed = np.asarray(
        cost_transformed,
        dtype=np.float64
    )

    cost_explainer = shap.TreeExplainer(
        cost_model.named_steps['classifier']
    )

    cost_values = cost_explainer.shap_values(
        cost_transformed
    )

    # Class 1 = Cost Overrun YES
    cost_values = cost_values[0, :, 1]

    cost_names = cost_model.named_steps[
        'preprocessor'
    ].get_feature_names_out()

    cost_df = pd.DataFrame({
        "Feature": cost_names,
        "SHAP Value": cost_values
    })

    cost_df["Clean Feature"] = cost_df["Feature"].apply(
        clean_feature_name
    )

    cost_df["Importance"] = cost_df["SHAP Value"].abs()

    # ---------- DELAY ----------
    delay_transformed = delay_model.named_steps[
        'preprocessor'
    ].transform(input_data)

    if hasattr(delay_transformed, "toarray"):
        delay_transformed = delay_transformed.toarray()

    delay_transformed = np.asarray(
        delay_transformed,
        dtype=np.float64
    )

    delay_explainer = shap.TreeExplainer(
        delay_model.named_steps['classifier']
    )

    delay_values = delay_explainer.shap_values(
        delay_transformed
    )

    # Class 1 = Delay YES
    delay_values = delay_values[0, :, 1]

    delay_names = delay_model.named_steps[
        'preprocessor'
    ].get_feature_names_out()

    delay_df = pd.DataFrame({
        "Feature": delay_names,
        "SHAP Value": delay_values
    })

    delay_df["Clean Feature"] = delay_df["Feature"].apply(
        clean_feature_name
    )

    delay_df["Importance"] = delay_df["SHAP Value"].abs()

    # ---------- SEPARATE RISK DRIVERS ----------

    cost_increasing = (
        cost_df[cost_df["SHAP Value"] > 0]
        .sort_values("SHAP Value", ascending=False)
        .head(5)
    )

    cost_reducing = (
        cost_df[cost_df["SHAP Value"] < 0]
        .sort_values("SHAP Value", ascending=True)
        .head(5)
    )

    delay_increasing = (
        delay_df[delay_df["SHAP Value"] > 0]
        .sort_values("SHAP Value", ascending=False)
        .head(5)
    )

    delay_reducing = (
        delay_df[delay_df["SHAP Value"] < 0]
        .sort_values("SHAP Value", ascending=True)
        .head(5)
    )

    return {
        "Cost Risk Drivers": cost_increasing,
        "Cost Risk Reducing Factors": cost_reducing,
        "Delay Risk Drivers": delay_increasing,
        "Delay Risk Reducing Factors": delay_reducing
    }
```
